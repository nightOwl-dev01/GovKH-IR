## 2. Khmer normalization and segmentation

Khmer spaces do not reliably delimit every word. A segmenter predicts word boundaries; a Transformer tokenizer then maps text into the model's own subword/token vocabulary. These are distinct operations.

This notebook preserves several text variants:
- `raw`: original text
- `normalized`: Unicode NFC + whitespace cleanup
- `segmented`: output from the selected Khmer word segmenter

The baseline segmenter below **only splits on existing spaces**. It is not a Khmer word segmenter and must not be reported as one.

```python
import re
import unicodedata

def normalize_text(text):
    """Normalize Unicode Khmer/English text without deleting ordinary letters."""
    text = unicodedata.normalize("NFC", str(text))
    text = text.replace("\\u200b", "").replace("\\ufeff", "")
    text = re.sub(r"[\\t\\r\\n]+", " ", text)
    text = re.sub(r"\\s+", " ", text).strip()
    return text

# Regression checks: English letters must remain intact; whitespace must collapse.
assert normalize_text("United Nations") == "United Nations"
assert normalize_text("Khmer\\tIR\\nResearch") == "Khmer IR Research"
assert normalize_text("  Hello   World  ") == "Hello World"

docs["normalized"] = docs["text"].map(normalize_text)
queries["normalized"] = queries["query"].map(normalize_text)

def whitespace_baseline(text):
    # Baseline only: preserves existing whitespace boundaries.
    return " ".join(normalize_text(text).split())

docs["segmented_whitespace"] = docs["normalized"].map(whitespace_baseline)
queries["segmented_whitespace"] = queries["normalized"].map(whitespace_baseline)

print("Normalization regression tests passed.")
display(docs[["doc_id", "text", "normalized"]].head())
```

### 2A. Optional segmenter integration

UnifiedCut, KhPOS and other Khmer segmenters may have changing package/repository interfaces. Rather than silently installing an unverified wrapper, this cell supports:
- using a segmenter you install and call yourself;
- supplying a CSV with pre-segmented text;
- plugging a Python callable into `segmenter_fn`.

If a project offers a current Colab installation command, follow its official README, then define `segmenter_fn(text) -> segmented_string`.

```python
# Select one of: "whitespace", "khmernlp", "custom"
# khmernlp is an available third-party Khmer tokenizer. Validate its output on your gold set.
SEGMENTER = "whitespace"

def whitespace_baseline(text):
    # Not true Khmer word segmentation; only splits on spaces already present.
    return " ".join(normalize_text(text).split())

def khmernlp_segment(text):
    # Optional package: pip install khmernlp
    # Its public usage exposes khmernlp.tokenize([...]); output format can vary by release.
    import khmernlp
    result = khmernlp.tokenize([normalize_text(text)])
    item = result[0] if isinstance(result, (list, tuple)) and len(result) else result
    if isinstance(item, (list, tuple)):
        return " ".join(map(str, item))
    return str(item)

def segmenter_fn(text):
    if SEGMENTER == "whitespace":
        return whitespace_baseline(text)
    elif SEGMENTER == "khmernlp":
        return khmernlp_segment(text)
    else:
        raise NotImplementedError("Set your custom segmenter_fn implementation here.")

segmenter_name = SEGMENTER
docs["segmented"] = docs["normalized"].map(segmenter_fn)
queries["segmented"] = queries["normalized"].map(segmenter_fn)

print("Selected segmenter:", segmenter_name)
display(docs[["doc_id", "normalized", "segmented"]].head())
```

### 2B. Segmentation evaluation (gold-standard boundaries required)

For a proper segmentation score, supply a gold tokenized text for each sentence/document. The evaluator compares **character boundary positions**, which avoids depending on token string formatting.

Upload `segmentation_gold.csv` with:
- `text`: original unsegmented Khmer text
- `gold_segmented`: gold tokens separated by spaces

Do not use machine-generated segmentation as gold. The demo below stays disabled until a gold file is supplied.

```python
def boundary_positions(segmented_text):
    # Boundary = character offset after each token, excluding the final end-of-string.
    tokens = [t for t in str(segmented_text).split() if t]
    positions, offset = set(), 0
    for tok in tokens[:-1]:
        offset += len(tok)
        positions.add(offset)
    return positions

def segmentation_boundary_metrics(gold, pred):
    g, p = boundary_positions(gold), boundary_positions(pred)
    tp = len(g & p)
    precision = tp / len(p) if p else (1.0 if not g else 0.0)
    recall = tp / len(g) if g else (1.0 if not p else 0.0)
    f1 = 2*precision*recall/(precision+recall) if precision+recall else 0.0
    return {"precision": precision, "recall": recall, "f1": f1, "gold_boundaries": len(g), "pred_boundaries": len(p)}

RUN_SEGMENTATION_GOLD_EVAL = False
if RUN_SEGMENTATION_GOLD_EVAL:
    from google.colab import files
    up = files.upload()
    gold_df = pd.read_csv(next(iter(up))).fillna("")
    seg_rows = [segmentation_boundary_metrics(r.gold_segmented, segmenter_fn(r.text))
                for r in gold_df.itertuples()]
    display(pd.DataFrame(seg_rows).describe())
```

## 3. Lexical retrieval: TF-IDF and BM25

Two tokenization settings are tested:
- whitespace tokens (baseline)
- segmented tokens (replace with a real segmenter for meaningful Khmer experiments)

TF-IDF uses sparse term-weight vectors. BM25 scores documents based on query-term matches, with term-frequency saturation and document-length normalization.

```python
def lexical_tokens(text):
    # For Khmer, this relies on whitespace segmentation. Replace with a validated segmenter.
    return [t for t in normalize_text(text).split() if t]

def fit_tfidf_retriever(doc_texts):
    vectorizer = TfidfVectorizer(tokenizer=lexical_tokens, token_pattern=None, lowercase=False)
    matrix = vectorizer.fit_transform(doc_texts)
    return vectorizer, matrix

def fit_bm25_retriever(doc_texts):
    tokenized = [lexical_tokens(t) for t in doc_texts]
    return BM25Okapi(tokenized)

def retrieve_tfidf(query, vectorizer, matrix, doc_ids, top_k=10):
    qv = vectorizer.transform([query])
    scores = cosine_similarity(qv, matrix).ravel()
    order = np.argsort(-scores)[:top_k]
    return [(doc_ids[i], float(scores[i])) for i in order]

def retrieve_bm25(query, bm25, doc_ids, top_k=10):
    scores = np.asarray(bm25.get_scores(lexical_tokens(query)))
    order = np.argsort(-scores)[:top_k]
    return [(doc_ids[i], float(scores[i])) for i in order]

doc_ids = docs["doc_id"].astype(str).tolist()
doc_texts_raw = docs["normalized"].tolist()
doc_texts_seg = docs["segmented"].tolist()

tfidf_raw, tfidf_raw_matrix = fit_tfidf_retriever(doc_texts_raw)
bm25_raw = fit_bm25_retriever(doc_texts_raw)
tfidf_seg, tfidf_seg_matrix = fit_tfidf_retriever(doc_texts_seg)
bm25_seg = fit_bm25_retriever(doc_texts_seg)

print("Lexical indexes built.")
```

## 4. Retrieval evaluation metrics

The notebook uses:
- **Recall@K:** fraction of relevant documents retrieved in the first K.
- **Precision@K:** fraction of the first K results that are relevant.
- **MRR:** reciprocal rank of the first relevant result.
- **nDCG@K:** ranking quality with higher reward for relevant items near the top.

For binary relevance labels, these are suitable standard IR metrics. For graded relevance, extend the evaluator to accept relevance grades.

```python
def parse_relevant(value):
    if isinstance(value, (list, tuple, set)):
        return set(map(str, value))
    return {x.strip() for x in str(value).split(";") if x.strip() and x.strip().lower() != "nan"}

def dcg(rels):
    return sum((2**rel - 1) / math.log2(i + 2) for i, rel in enumerate(rels))

def score_ranking(ranked_ids, relevant_ids, k=10):
    relevant_ids = set(relevant_ids)
    top = list(ranked_ids[:k])
    hits = [1 if x in relevant_ids else 0 for x in top]
    recall = sum(hits) / len(relevant_ids) if relevant_ids else np.nan
    precision = sum(hits) / k if k else np.nan
    rr = 0.0
    for rank, hit in enumerate(hits, start=1):
        if hit:
            rr = 1.0 / rank
            break
    ideal_hits = min(len(relevant_ids), k)
    ideal = [1] * ideal_hits + [0] * (k - ideal_hits)
    idcg = dcg(ideal)
    ndcg = dcg(hits) / idcg if idcg else 0.0
    return {"recall": recall, "precision": precision, "mrr": rr, "ndcg": ndcg}

def evaluate_retriever(name, retrieval_fn, query_df=queries, k=5, text_column="normalized"):
    rows = []
    for r in query_df.itertuples():
        result = retrieval_fn(getattr(r, text_column), k)
        ranked = [x[0] for x in result]
        relevant = parse_relevant(r.relevant_doc_ids)
        m = score_ranking(ranked, relevant, k)
        rows.append({"system": name, "query_id": r.query_id, **m})
    return pd.DataFrame(rows)

results_by_query = []
results_by_query.append(evaluate_retriever(
    "TF-IDF / normalized", lambda q,k: retrieve_tfidf(q, tfidf_raw, tfidf_raw_matrix, doc_ids, k)))
results_by_query.append(evaluate_retriever(
    "BM25 / normalized", lambda q,k: retrieve_bm25(q, bm25_raw, doc_ids, k)))
results_by_query.append(evaluate_retriever(
    "TF-IDF / segmented", lambda q,k: retrieve_tfidf(q, tfidf_seg, tfidf_seg_matrix, doc_ids, k)))
results_by_query.append(evaluate_retriever(
    "BM25 / segmented", lambda q,k: retrieve_bm25(q, bm25_seg, doc_ids, k)))

lexical_results = pd.concat(results_by_query, ignore_index=True)
display(lexical_results.groupby("system")[["recall","precision","mrr","ndcg"]].mean().round(3))
```

## 5. Dense embeddings: EmbeddingGemma 2 and Qwen3-VL-Embedding-2B

### Important implementation note

Embedding model repositories and APIs can change. The cells below are deliberately modular:
- set `ENABLE_DENSE_MODELS = True` only when you are ready to download model weights;
- use each model's official model card / repository for its recommended loading and encoding API;
- if an architecture-specific model class is not available in your installed Transformers version, the cell reports the issue rather than breaking lexical experiments.

**Do not assume equal vector dimensions imply compatible embedding spaces.** Never compare a query from one model against document vectors from another model. Each model needs its own document index and query encoder.

```python
ENABLE_DENSE_MODELS = False  # Change to True when ready to download model weights.
EMBEDDINGGEMMA_MODEL_ID = "google/embeddinggemma-2"
QWEN_VL_EMBED_MODEL_ID = "Qwen/Qwen3-VL-Embedding-2B"

dense_models = {}
dense_errors = {}

if ENABLE_DENSE_MODELS:
    import torch
    from sentence_transformers import SentenceTransformer
    device = "cuda" if torch.cuda.is_available() else "cpu"
    print("Device:", device)

    for model_name, model_id in [
        ("EmbeddingGemma2", EMBEDDINGGEMMA_MODEL_ID),
        ("Qwen3-VL-Embedding-2B", QWEN_VL_EMBED_MODEL_ID)
    ]:
        try:
            # Official SentenceTransformer interface. This is preferable to generic
            # AutoModel + guessed pooling because the models have their own recipes.
            model = SentenceTransformer(
                model_id,
                device=device,
                trust_remote_code=True
            )
            dense_models[model_name] = model
            print("Loaded:", model_name)
        except Exception as e:
            dense_errors[model_name] = repr(e)
            print("Could not load", model_name, ":", repr(e))
else:
    print("Dense model loading is disabled. Lexical tests can run now.")
```

### 5A. Generic text embedding adapter

Different model repositories can require different prompt formatting, pooling, image preprocessing, or custom code. This adapter is intentionally a **baseline only**. Check the official model card and adjust the model-specific encoding function before publishing results.

```python
def encode_texts(model, texts, batch_size=8, prompt=None):
    kwargs = {
        "batch_size": batch_size,
        "convert_to_numpy": True,
        "normalize_embeddings": True,
        "show_progress_bar": True,
    }
    if prompt is not None:
        kwargs["prompt"] = prompt
    vectors = model.encode(list(texts), **kwargs)
    return np.asarray(vectors, dtype=np.float32)

dense_doc_vectors = {}
dense_query_vectors = {}

if dense_models:
    for name, model in dense_models.items():
        try:
            # Qwen's default prompt is "Represent the user's input.";
            # use a retrieval-specific query instruction for query embeddings.
            if name == "Qwen3-VL-Embedding-2B":
                qv = model.encode(
                    queries["normalized"].tolist(),
                    prompt="Retrieve relevant documents for the user's query.",
                    normalize_embeddings=True,
                    convert_to_numpy=True,
                    show_progress_bar=True
                )
            else:
                qv = model.encode(
                    queries["normalized"].tolist(),
                    normalize_embeddings=True,
                    convert_to_numpy=True,
                    show_progress_bar=True
                )
            dv = model.encode(
                docs["normalized"].tolist(),
                normalize_embeddings=True,
                convert_to_numpy=True,
                show_progress_bar=True
            )
            dense_doc_vectors[name] = np.asarray(dv, dtype=np.float32)
            dense_query_vectors[name] = np.asarray(qv, dtype=np.float32)
            print(name, "documents:", dense_doc_vectors[name].shape,
                  "queries:", dense_query_vectors[name].shape)
        except Exception as e:
            print("Encoding failed for", name, ":", repr(e))
```

### 5B. Dense retrieval and optional raw-vs-segmented input

The generic adapter above is not guaranteed to reproduce the official pooling/instruction recipe for every model. Treat its scores as provisional until you use the exact recommended model API.

```python
def retrieve_dense(query_vector, doc_vectors, doc_ids, top_k=10):
    q = np.asarray(query_vector).reshape(1, -1)
    D = np.asarray(doc_vectors)
    scores = cosine_similarity(q, D).ravel()
    order = np.argsort(-scores)[:top_k]
    return [(doc_ids[i], float(scores[i])) for i in order]

dense_result_frames = []
if dense_doc_vectors and dense_query_vectors:
    for model_name in dense_doc_vectors:
        dv, qv = dense_doc_vectors[model_name], dense_query_vectors[model_name]
        rows = []
        for i, qrow in enumerate(queries.itertuples()):
            ranked = [x[0] for x in retrieve_dense(qv[i], dv, doc_ids, 5)]
            rows.append({"system": model_name + " / normalized", "query_id": qrow.query_id,
                         **score_ranking(ranked, parse_relevant(qrow.relevant_doc_ids), 5)})
        dense_result_frames.append(pd.DataFrame(rows))

if dense_result_frames:
    dense_results = pd.concat(dense_result_frames, ignore_index=True)
    all_results = pd.concat([lexical_results, dense_results], ignore_index=True)
else:
    all_results = lexical_results.copy()
    print("No dense results yet; showing lexical baselines.")
```

### 5C. Optional segmented-input dense comparison

This directly tests whether Khmer word segmentation helps or hurts each embedding model. Keep the same corpus and query labels; only change the text preparation.

```python
segmented_dense_results = []
if dense_models:
    for name, model in dense_models.items():
        try:
            if name == "Qwen3-VL-Embedding-2B":
                qv_seg = model.encode(
                    queries["segmented"].tolist(),
                    prompt="Retrieve relevant documents for the user's query.",
                    normalize_embeddings=True, convert_to_numpy=True
                )
            else:
                qv_seg = model.encode(
                    queries["segmented"].tolist(),
                    normalize_embeddings=True, convert_to_numpy=True
                )
            dv_seg = model.encode(
                docs["segmented"].tolist(),
                normalize_embeddings=True, convert_to_numpy=True
            )
            for qi, qrow in enumerate(queries.itertuples()):
                ranked = [x[0] for x in retrieve_dense(qv_seg[qi], dv_seg, doc_ids, 5)]
                segmented_dense_results.append({
                    "system": name + " / segmented",
                    "query_id": qrow.query_id,
                    **score_ranking(ranked, parse_relevant(qrow.relevant_doc_ids), 5)
                })
        except Exception as e:
            print("Segmented dense experiment failed for", name, ":", repr(e))

if segmented_dense_results:
    all_results = pd.concat([all_results, pd.DataFrame(segmented_dense_results)], ignore_index=True)
```

## 6. Hybrid retrieval with Reciprocal Rank Fusion (RRF)

RRF combines rankings without requiring the lexical and dense systems to have comparable raw score scales.

\[
RRF(d)=\sum_{r \in systems}\frac{1}{k_0 + rank_r(d)}
\]

Here `k0=60` is a common starting value. Tune it only on a validation set, not on the final test set.

```python
def rrf_fuse(rankings, k0=60, top_k=10):
    scores = defaultdict(float)
    for ranked_ids in rankings:
        for rank, doc_id in enumerate(ranked_ids, start=1):
            scores[doc_id] += 1.0 / (k0 + rank)
    return sorted(scores.items(), key=lambda x: x[1], reverse=True)[:top_k]

hybrid_rows = []
for qi, qrow in enumerate(queries.itertuples()):
    lexical_rank = [x[0] for x in retrieve_bm25(qrow.normalized, bm25_raw, doc_ids, len(doc_ids))]
    ranking_list = [lexical_rank]
    for model_name, qv in dense_query_vectors.items():
        ranking_list.append([x[0] for x in retrieve_dense(qv[qi], dense_doc_vectors[model_name], doc_ids, len(doc_ids))])
    if len(ranking_list) > 1:
        fused = [x[0] for x in rrf_fuse(ranking_list, top_k=5)]
        hybrid_rows.append({"system": "Hybrid RRF", "query_id": qrow.query_id,
                            **score_ranking(fused, parse_relevant(qrow.relevant_doc_ids), 5)})

if hybrid_rows:
    all_results = pd.concat([all_results, pd.DataFrame(hybrid_rows)], ignore_index=True)
```

### 7B. EmbeddingGemma 2 Matryoshka dimension experiment

If EmbeddingGemma 2 was loaded and encoded at 768 dimensions, test its supported truncation sizes. Always re-normalize after truncation. Do not apply these dimensions to Qwen unless its model card explicitly supports them.

```python
if "EmbeddingGemma2" in dense_doc_vectors:
    full_docs = dense_doc_vectors["EmbeddingGemma2"]
    full_queries = dense_query_vectors["EmbeddingGemma2"]
    if full_docs.shape[1] == 768:
        dim_rows = []
        for dim in [128, 256, 512, 768]:
            dvec = full_docs[:, :dim]
            qvec = full_queries[:, :dim]
            dvec = dvec / np.clip(np.linalg.norm(dvec, axis=1, keepdims=True), 1e-12, None)
            qvec = qvec / np.clip(np.linalg.norm(qvec, axis=1, keepdims=True), 1e-12, None)
            per_q = []
            for qi, qrow in enumerate(queries.itertuples()):
                ranked = [x[0] for x in retrieve_dense(qvec[qi], dvec, doc_ids, 5)]
                per_q.append(score_ranking(ranked, parse_relevant(qrow.relevant_doc_ids), 5))
            dim_rows.append({"dimensions":dim, **pd.DataFrame(per_q).mean().to_dict()})
        dim_results = pd.DataFrame(dim_rows)
        display(dim_results)
        plt.figure(figsize=(8,4))
        sns.lineplot(data=dim_results, x="dimensions", y="ndcg", marker="o", label="nDCG@5")
        sns.lineplot(data=dim_results, x="dimensions", y="recall", marker="o", label="Recall@5")
        plt.title("EmbeddingGemma 2: retrieval quality vs dimension")
        plt.ylim(0,1); plt.grid(alpha=.25); plt.show()
    else:
        print("EmbeddingGemma output is not 768D; got", full_docs.shape)
else:
    print("Load EmbeddingGemma 2 first to run the MRL dimension experiment.")
```

## 8. Embedding dimension and vector storage

Storage estimate for dense float32 vectors:

\[
storage \approx N_{documents} \times dimensions \times 4\ bytes
\]

This excludes vector-index overhead, metadata, replicas, and compression.

```python
def vector_storage_mb(n_vectors, dimensions, bytes_per_value=4):
    return n_vectors * dimensions * bytes_per_value / (1024**2)

N_DOCS_FOR_ESTIMATE = len(docs)
dimension_options = [128, 256, 384, 512, 768, 1024, 1536, 2048, 2560, 3072]
storage_df = pd.DataFrame({
    "dimensions": dimension_options,
    "storage_MB": [vector_storage_mb(N_DOCS_FOR_ESTIMATE, d) for d in dimension_options],
    "storage_GB_for_1M": [vector_storage_mb(1_000_000, d)/1024 for d in dimension_options]
})
display(storage_df.round(3))
plt.figure(figsize=(9,4))
sns.lineplot(data=storage_df, x="dimensions", y="storage_GB_for_1M", marker="o")
plt.title("Float32 vector storage for 1 million documents")
plt.ylabel("Storage (GiB)")
plt.xlabel("Embedding dimensions")
plt.grid(alpha=.25)
plt.show()
```

```python
# Latency benchmark for lexical systems. Run several repetitions for stable measurements.
def time_retriever(fn, qlist, repeats=3):
    samples = []
    for _ in range(repeats):
        t0 = time.perf_counter()
        for q in qlist:
            fn(q, 5)
        samples.append((time.perf_counter()-t0)/max(1,len(qlist))*1000)
    return float(np.median(samples))

latency_rows = [
    {"system":"TF-IDF / normalized", "ms_per_query":time_retriever(lambda q,k: retrieve_tfidf(q,tfidf_raw,tfidf_raw_matrix,doc_ids,k), queries.normalized.tolist())},
    {"system":"BM25 / normalized", "ms_per_query":time_retriever(lambda q,k: retrieve_bm25(q,bm25_raw,doc_ids,k), queries.normalized.tolist())},
    {"system":"TF-IDF / segmented", "ms_per_query":time_retriever(lambda q,k: retrieve_tfidf(q,tfidf_seg,tfidf_seg_matrix,doc_ids,k), queries.normalized.tolist())},
    {"system":"BM25 / segmented", "ms_per_query":time_retriever(lambda q,k: retrieve_bm25(q,bm25_seg,doc_ids,k), queries.normalized.tolist())}
]
latency_df = pd.DataFrame(latency_rows)
display(latency_df)
plt.figure(figsize=(9,4))
sns.barplot(data=latency_df, x="system", y="ms_per_query")
plt.title("Lexical retrieval latency")
plt.ylabel("Milliseconds per query")
plt.xticks(rotation=25, ha="right")
plt.tight_layout()
plt.show()
```

## 10. Error analysis: inspect failed queries

This section shows the ranked results for each query and flags cases where no relevant document appears in the top five. This is especially useful for diagnosing:
- Khmer word-boundary errors;
- spelling variants and OCR noise;
- cross-language query/document mismatch;
- semantic drift;
- missing or incomplete relevance judgments.

```python
def show_query_results(query_id, system="BM25 / normalized", top_k=5):
    qrow = queries.loc[queries.query_id.astype(str) == str(query_id)].iloc[0]
    relevant = parse_relevant(qrow["relevant_doc_ids"])
    qtext = qrow["normalized"]
    if system == "TF-IDF / normalized":
        ranked = retrieve_tfidf(qtext, tfidf_raw, tfidf_raw_matrix, doc_ids, top_k)
    elif system == "BM25 / normalized":
        ranked = retrieve_bm25(qtext, bm25_raw, doc_ids, top_k)
    elif system == "TF-IDF / segmented":
        ranked = retrieve_tfidf(qrow["segmented"], tfidf_seg, tfidf_seg_matrix, doc_ids, top_k)
    elif system == "BM25 / segmented":
        ranked = retrieve_bm25(qrow["segmented"], bm25_seg, doc_ids, top_k)
    elif system in dense_doc_vectors:
        qi = queries.index[queries.query_id.astype(str) == str(query_id)][0]
        ranked = retrieve_dense(dense_query_vectors[system][qi], dense_doc_vectors[system], doc_ids, top_k)
    else:
        raise ValueError("Unknown system: " + system)
    out = []
    for rank, (did, score) in enumerate(ranked, start=1):
        row = docs.loc[docs.doc_id.astype(str) == str(did)].iloc[0]
        out.append({"rank":rank, "doc_id":did, "score":score, "relevant":did in relevant, "text":row["text"]})
    return pd.DataFrame(out)

display(show_query_results(str(queries.iloc[0].query_id)))
```
